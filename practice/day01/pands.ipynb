```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ana", "Luis", "Ana", "Sofi", None],
    "city": ["BA", "Cordoba", "BA", "Rosario", "BA"],
    "salary": [3000, 4500, 3000, None, 2800],
})

df.shape
```

```python
df["bonus"] = df["salary"] * 0.1
df.head()
```

```python
df = df.rename(columns={"salary": "wage"})
df.head()
```

```python
df = df.drop_duplicates()
df.head()
```

```python
df = df.dropna(subset=["name"])
df.head()
```

```python
df["wage"] = df["wage"].fillna(0)
df.head() 
```

```python
df["wage"] = df["wage"].astype(int)
df.head()
```

```python
df.sort_values("wage", ascending=False)
df.head()
```

```python
df.groupby("city")["wage"].mean()
```

```python
df.groupby("city").agg(n=("name", "count"), avg=("wage", "mean"))
```
